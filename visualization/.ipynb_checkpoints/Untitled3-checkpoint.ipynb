```python
import pandas as pd
df=pd.read_csv('/home/punit/Downloads/Mint/soptify/datase/clean_data.csv')
df
```

```python
import seaborn as sea
import matplotlib.pyplot as plt

plt.figure(figsize=(14,10))
sea.barplot(data=df,x='popularity',y='duration_ms')
plt.show()
```
