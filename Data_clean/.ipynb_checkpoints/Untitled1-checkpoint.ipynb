```python
import pandas as pd
df=pd.read_csv('/home/punit/Downloads/Mint/soptify/datase/clean_data.csv')
df
```

```python
en_piv=pd.pivot_table(data=df,values=['energy','popularity'],index='track_id')
en_piv
```

```python
import matplotlib.pyplot as plt
import seaborn as sea
plt.figure(figsize=(14,10))
sea.regplot(data=en_piv,x='popularity',y='energy')
plt.show()
```
