```python
import seaborn as sea
import matplotlib.pyplot as plt

sea.set_theme('notebook')
sea.set_style('whitegrid'
plt.figure(figsize=(14,10))
sea.histplot(data=df['popularity'],bins=20,kde=True)
```

```python
import seaborn as sea
import matplotlib.pyplot as plt
df_pop=df[['popularity','duration_ms']]
plt.figure(figsize=(14,10))
sea.kdeplot(data=df_pop,x='popularity')
plt.show()
```

```python

plt.figure(figsize=(14,10))

df_pop=df[['popularity','duration_ms']]
sea.barplot(data=df_pop,x='popularity',y='duration_ms')
plt.xticks(rotation=90)
plt.show()
```
