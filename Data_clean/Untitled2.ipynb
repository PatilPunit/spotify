```python
import pandas as pd
df=pd.read_csv('/home/punit/Downloads/Mint/soptify/datase/clean_data.csv')
df
```

```python
import seaborn as sea
import matplotlib.pyplot as plt

sea.set_theme('notebook')
sea.set_style('whitegrid'
plt.figure(figsize=(14,10))
sea.histplot(data=df['popularity'],bins=20,kde=True)
```

```python
import seaborn as sea
import matplotlib.pyplot as plt
df_pop=df[['popularity','duration_ms']]
plt.figure(figsize=(14,10))
sea.kdeplot(data=df_pop,x='popularity')
plt.show()
```

```python

plt.figure(figsize=(14,10))

df_pop=df[['popularity','duration_ms']]
sea.barplot(data=df_pop,x='popularity',y='duration_ms')
plt.xticks(rotation=90)
plt.show()
```

```python

plt.figure(figsize=(14,10))
df_pop=df[['popularity','danceability']]
sea.lineplot(data=df_pop,x='popularity',y='danceability')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(14,10))
sea.countplot(
    data=df, 
    y='track_genre', 
    color='hotpink', 
    order=df['track_genre'].value_counts().index
)
plt.xticks(rotation=90)
plt.show()
```

```python
popular_80 = df[df['popularity'] > 80]

genre_count = popular_80['track_genre'].value_counts().head(20)
plt.figure(figsize=(12,6))
sea.barplot(x=genre_count.values,
            y=genre_count.index)

plt.title("Genres with Most Tracks (Popularity > 80)")
plt.show()
```
