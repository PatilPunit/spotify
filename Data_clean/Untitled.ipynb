```python
import pandas as pd
df=pd.read_csv('/home/punit/Downloads/Mint/soptify/datase/clean_data.csv')
df
```

```python
import matplotlib.pyplot as plt
import  seaborn as sea

plt.figure(figsize=(14,10))
sea.barplot(data=df,x='popularity',y='danceability')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(14,10))
sea.scatterplot(data=df,x='popularity',y='danceability')
plt.show()
```

```python
dance=df[['popularity','danceability']]
top_dan=dance.sort_values(ascending=False,by='popularity')
top_dan
```

```python
avg_dan_50=top_dan.head(50).mean()
avg_dan_100=top_dan.head(100).mean()
avg_dan_200=top_dan.head(200).mean()
print(avg_dan_50)
print('______________________________________')
print(avg_dan_100)
print('______________________________________')
print(avg_dan_200)
```

```python
avg_dan_50=top_dan.tail(50).mean()
avg_dan_100=top_dan.tail(100).mean()
avg_dan_200=top_dan.tail(200).mean()
print(avg_dan_50)
print('______________________________________')
print(avg_dan_100)
print('________________________________________')
print(avg_dan_200)
```
