```python
import pandas as pd
df = pd.read_csv('/home/punit/Documents/soptify/datase/no_outlier.csv')
df
```

```python
import seaborn as sea
import matplotlib.pyplot as plt

sea.set_theme('notebook')
sea.set_style('whitegrid')
plt.figure(figsize=(14,10))

df_pop=df[['popularity','duration_ms']]
sea.barplot(data=df_pop,x='popularity',y='duration_ms')
plt.xticks(rotation=90)
plt.show()
```
